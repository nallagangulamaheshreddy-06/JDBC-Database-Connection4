```python

import javafx.application.Application;
import javafx.geometry.Insets;
import javafx.scene.Scene;
import javafx.scene.control.*;
import javafx.scene.layout.VBox;
import javafx.stage.Stage;

public class TipCalculator extends Application {

    public void start(Stage stage) {

        Label title = new Label("Tip Calculator");

        Label billLabel = new Label("Enter Bill Amount:");
        TextField billField = new TextField();

        Label tipLabel = new Label("Select Tip Percentage:");
        ComboBox<Integer> tipBox = new ComboBox<>();
        tipBox.getItems().addAll(5, 10, 15, 20, 25);
        tipBox.setValue(10);

        Label peopleLabel = new Label("Number of People:");
        TextField peopleField = new TextField();
        peopleField.setText("1");

        Button calculateButton = new Button("Calculate");

        Label result = new Label();

        calculateButton.setOnAction(e -> {
            try {
                double bill = Double.parseDouble(billField.getText());
                int tip = tipBox.getValue();
                int people = Integer.parseInt(peopleField.getText());

                if (bill < 0 || people <= 0) {
                    result.setText("Enter valid values!");
                    return;
                }

                double tipAmount = bill * tip / 100;
                double total = bill + tipAmount;
                double perPerson = total / people;

                result.setText(
                    "Tip Amount: Rs. " + tipAmount +
                    "\nTotal Bill: Rs. " + total +
                    "\nPer Person: Rs. " + perPerson
                );

            } catch (NumberFormatException ex) {
                result.setText("Please enter valid numbers!");
            }
        });

        VBox root = new VBox(10);
        root.setPadding(new Insets(20));

        root.getChildren().addAll(
            title, billLabel, billField,
            tipLabel, tipBox,
            peopleLabel, peopleField,
            calculateButton, result
        );

        Scene scene = new Scene(root, 350, 400);

        stage.setTitle("Tip Calculator");
        stage.setScene(scene);
        stage.show();
    }

    public static void main(String[] args) {
        launch(args);
    }
}
```
